```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()
DATA_PATH = PROJECT_ROOT / "data_raw"

train_raw = pd.read_csv(DATA_PATH / "train.csv")
test_raw = pd.read_csv(DATA_PATH / "test.csv")

print("Thư mục:", PROJECT_ROOT)
print("Train:", train_raw.shape)
print("Test:", test_raw.shape)

print("\nPhân bố nhãn train:")
print(train_raw["kredit"].value_counts().sort_index())
display(train_raw.head(3))
```

```python

# 0 = Good, 1 = Bad
bad_credit = 1 - train_raw["kredit"]

print(bad_credit.value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split

train_data = train_raw.drop(columns=["kredit"]).copy()
train_data["bad_credit"] = bad_credit

development, benchmark = train_test_split(
    train_data,
    test_size=0.20,
    random_state=42,
    stratify=train_data["bad_credit"]
)

development = development.sort_values("Id").reset_index(drop=True)
benchmark = benchmark.sort_values("Id").reset_index(drop=True)

competition = test_raw.copy()

for name, data in [
    ("Development", development),
    ("Benchmark", benchmark)
]:
    print(name, data.shape)
    print(data["bad_credit"].value_counts().sort_index())
    print()

print("Competition:", competition.shape)
```

```python
OUTPUT_PATH = PROJECT_ROOT / "data_splits_colab" / "data_split"
OUTPUT_PATH.mkdir(parents=True, exist_ok=True)

development.to_csv(
    OUTPUT_PATH / "development_640.csv",
    index=False
)

benchmark.to_csv(
    OUTPUT_PATH / "historical_benchmark_160.csv",
    index=False
)

competition.to_csv(
    OUTPUT_PATH / "competition_200.csv",
    index=False
)

print("Đã lưu dữ liệu vào:", OUTPUT_PATH)
```
